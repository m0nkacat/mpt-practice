<h2>Файл: maps_scores.csv — карты FNATIC</h2>
<p>Счёт на каждой карте.</p>
<ul>
    <li><code>Tournament</code>, <code>Stage</code>, <code>Match Type</code>, <code>Match Name</code>, <code>Map</code> — турнир, стадия, тип матча, матч, карта;</li>
    <li><code>Team A</code>, <code>Team B</code> — команды;</li>
    <li><code>Team A Score</code>, <code>Team B Score</code> — раунды, выигранные командой;</li>
    <li><code>... Attacker Score</code>, <code>... Defender Score</code> — раунды, выигранные за атаку и за защиту;</li>
    <li><code>... Overtime Score</code> — раунды, выигранные в овертайме;</li>
    <li><code>Duration</code> — длительность карты.</li>
</ul>

```python
path = '../datasets/archive/vct_2025/matches/maps_scores.csv'
df = pd.read_csv(path)
df.info()
display(df.head())
print(df.shape)
print(df.shape[0]) # 0 - это строки
print(df.shape[1]) # 1 - это столбцы
display(df.describe())
display(df.isnull().sum())
print('Дубликатов:', df.duplicated().sum())
```

```python
df = df.replace('Mega Minors', 'NRG') # ошибка в исходных данных: NRG записана как Mega Minors
df.drop('Duration', axis = 1, inplace = True) # удаляем признак: есть пропуски и для вопроса не нужен
df[['Team A Overtime Score', 'Team B Overtime Score']] = df[['Team A Overtime Score', 'Team B Overtime Score']].fillna(0) # нет овертайма -> 0
df = df.sort_values(by = 'Tournament')
df = df.drop_duplicates()
display(df.head())
print(df.shape)
df.to_csv('../datasets/processed/maps_scores.csv')
```

<p>Процент побед FNATIC на каждой карте за сезон</p>

```python
fm = df[(df['Team A'] == 'FNATIC') | (df['Team B'] == 'FNATIC')].copy()
fm['FNATIC Score'] = np.where(fm['Team A'] == 'FNATIC', fm['Team A Score'], fm['Team B Score'])
fm['Opponent Score'] = np.where(fm['Team A'] == 'FNATIC', fm['Team B Score'], fm['Team A Score'])
fm['Win'] = fm['FNATIC Score'] > fm['Opponent Score']
map_stats = fm.groupby('Map')['Win'].agg(['sum', 'count'])
map_stats.columns = ['Побед', 'Сыграно']
map_stats['% побед'] = (map_stats['Побед'] / map_stats['Сыграно'] * 100).round(1)
display(map_stats.sort_values(by = '% побед'))
```

<p>Карты в проигранных гранд-финалах</p>

```python
gf_maps = fm[(fm['Match Type'] == 'Grand Final') & fm['Tournament'].isin(lost)]
display(gf_maps[['Tournament', 'Map', 'FNATIC Score', 'Opponent Score', 'Win']])
```

<p><b>Ответ:</b></p>
<ul>
    <li>Сильные карты FNATIC — Split и Icebox, слабые — Sunset, Ascent и Pearl.</li>
    <li>Sunset и Lotus были в обоих проигранных финалах и оба раза закончились поражением.</li>
</ul>

<h2>Файл: overview.csv — игроки FNATIC</h2>
<p>Основная статистика игрока на карте.</p>
<ul>
    <li><code>Tournament</code>, <code>Stage</code>, <code>Match Type</code>, <code>Match Name</code>, <code>Map</code>, <code>Player</code>, <code>Team</code> — турнир, стадия, тип матча, матч, карта, игрок, команда;</li>
    <li><code>Agents</code> — агент;</li>
    <li><code>Rating</code> — рейтинг игрока;</li>
    <li><code>Average Combat Score</code> (ACS) — средний боевой счёт за раунд;</li>
    <li><code>Kills</code>, <code>Deaths</code>, <code>Assists</code> — убийства, смерти, помощь;</li>
    <li><code>Kills - Deaths (KD)</code> — разница убийств и смертей;</li>
    <li><code>Kill, Assist, Trade, Survive %</code> (KAST) — процент раундов, где игрок убил, помог, был «разменян» или выжил;</li>
    <li><code>Average Damage Per Round</code> (ADR) — средний урон за раунд;</li>
    <li><code>Headshot %</code> — процент попаданий в голову;</li>
    <li><code>First Kills</code>, <code>First Deaths</code>, <code>Kills - Deaths (FKD)</code> — первые убийства и смерти в раунде и их разница;</li>
    <li><code>Side</code> — сторона: <code>both</code> (итог), <code>attack</code>, <code>defend</code>.</li>
</ul>

```python
path = '../datasets/archive/vct_2025/matches/overview.csv'
df = pd.read_csv(path)
df.info()
display(df.head())
print(df.shape)
print(df.shape[0]) # 0 - это строки
print(df.shape[1]) # 1 - это столбцы
display(df.describe())
display(df.isnull().sum())
print('Дубликатов:', df.duplicated().sum())
```

```python
df = df.replace('Mega Minors', 'NRG') # ошибка в исходных данных: NRG записана как Mega Minors
df = df[df['Side'] == 'both'] # оставляем итог по обеим сторонам, attack / defend удаляем
df.drop(['Side', 'Kills - Deaths (KD)', 'Kills - Deaths (FKD)'], axis = 1, inplace = True) # KD и FKD считаются из других столбцов
df['Kill, Assist, Trade, Survive %'] = df['Kill, Assist, Trade, Survive %'].str.replace('%', '').astype(float)
df['Headshot %'] = df['Headshot %'].str.replace('%', '').astype(float)
df = df[df['Map'] != 'All Maps'] # удаляем строки-итоги по всем картам
df = df.dropna()
df = df.sort_values(by = 'Rating', ascending = False)
df = df.drop_duplicates()
display(df.head())
print(df.shape)
df.to_csv('../datasets/processed/overview.csv')
```

<p>Количество записей по командам (целевая переменная Team)</p>

```python
display(df['Team'].value_counts().head(10))
```

<p>Средняя статистика игроков FNATIC: весь сезон и проигранные гранд-финалы</p>

```python
cols = ['Rating', 'Average Combat Score', 'Kills', 'Deaths', 'Average Damage Per Round', 'First Kills', 'First Deaths']
fnc = df[df['Team'] == 'FNATIC']
fnc_gf = fnc[(fnc['Match Type'] == 'Grand Final') & fnc['Tournament'].isin(lost)]
display(pd.DataFrame({'Весь сезон': fnc[cols].mean(), 'Проигранные гранд-финалы': fnc_gf[cols].mean()}).round(2))
```

</ul>

<p>Исход и способ окончания каждого раунда.</p>
<ul>
    <li><code>Tournament</code>, <code>Stage</code>, <code>Match Type</code>, <code>Match Name</code>, <code>Map</code>, <code>Round Number</code> — турнир, стадия, тип матча, матч, карта, номер раунда;</li>
    <li><code>Team</code> — команда;</li>
    <li><code>Method</code> — способ окончания раунда: <code>Elimination</code> / <code>Eliminated</code> (убили всех / всех убили), <code>Detonated</code> (взрыв спайка), <code>Defused</code> (спайк обезврежен), <code>Time Expiry</code> (вышло время) и противоположные им для проигравшей стороны;</li>

<h2>Файл: eco_rounds.csv — экономика FNATIC</h2>
<p>Экономика каждого раунда.</p>
<ul>
    <li><code>Tournament</code>, <code>Stage</code>, <code>Match Type</code>, <code>Match Name</code>, <code>Map</code> — турнир, стадия, тип матча, матч, карта;</li>

    <li><code>Team</code> — команда;</li>
    <li><code>Loadout Value</code> — стоимость снаряжения команды;</li>
    <li><code>Remaining Credits</code> — оставшиеся кредиты;</li>
    <li><code>Type</code> — тип закупки (эко, полу-эко, полу-закупка, полная закупка);</li>
    <li><code>Outcome</code> — исход раунда (Win / Loss).</li>
</ul>
<p>В этом файле нет данных по Champions 2025, поэтому экономику финала смотрим только для Masters Toronto. У Masters Toronto не заполнена стоимость снаряжения (<code>Loadout Value</code>), но тип закупки (<code>Type</code>) есть, поэтому эти строки оставляем.</p>

```python
path = '../datasets/archive/vct_2025/matches/eco_rounds.csv'
df = pd.read_csv(path)
df.info()
display(df.head())
print(df.shape)
print(df.shape[0]) # 0 - это строки
print(df.shape[1]) # 1 - это столбцы
display(df.describe())
display(df.isnull().sum())
print('Дубликатов:', df.duplicated().sum())
```

```python
df = df.replace('Mega Minors', 'NRG') # ошибка в исходных данных: NRG записана как Mega Minors
df['Loadout Value'] = df['Loadout Value'].str.replace('k', '').astype(float) * 1000 # '24.3k' -> 24300
df['Remaining Credits'] = df['Remaining Credits'].str.replace('k', '').astype(float) * 1000
# пропуски в Loadout Value не удаляем: это весь Masters Toronto, а тип закупки (Type) для него есть
df = df.sort_values(by = 'Round Number')
df = df.drop_duplicates()
display(df.head())
print(df.shape)
df.to_csv('../datasets/processed/eco_rounds.csv')
```
